```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download("punkt_tab", quiet=True)


# 1. INTENTS

topics = {
    "greeting": [
        "hi", "hello", "hey", "good morning",
        "good afternoon", "good evening", "how are you"
    ],

    "python": [
        "python", "programming language"
    ],

    "nlp": [
        "nlp", "natural language processing"
    ],

    "chatbot": [
        "chatbot", "chat bot", "bot"
    ]
}


# 2. RESPONSES

responses = {
    "greeting": {
        "positive": "Hello! Great to see your enthusiasm!",
        "negative": "Hello! I'm here to help you."
    },

    "python": {
        "positive": "Python is a great language for AI and programming!",
        "negative": "Don't worry! You can learn Python step by step."
    },

    "nlp": {
        "positive": "NLP is an interesting field of Artificial Intelligence!",
        "negative": "NLP may seem difficult, but practice makes it easier."
    },

    "chatbot": {
        "positive": "Chatbots are a great way to apply NLP concepts!",
        "negative": "Let's work through the chatbot problems step by step."
    }
}


# 3. SENTIMENT WORDS

positive_words = {
    "good", "great", "happy", "love", "excellent",
    "awesome", "amazing", "wonderful", "enjoy", "helpful"
}

negative_words = {
    "bad", "sad", "hate", "angry", "terrible",
    "awful", "boring", "upset", "poor", "difficult"
}

negation_words = {
    "not", "no", "never", "don't", "dont",
    "didn't", "didnt", "isn't", "isnt",
    "wasn't", "wasnt", "cannot"
}


# 4. SENTIMENT ANALYSIS

def analyse_sentiment(tokens):

    positive_count = 0
    negative_count = 0
    negate_next = False

    for word in tokens:

        if word in negation_words:
            negate_next = True
            continue

        if word in positive_words:

            if negate_next:
                negative_count += 1
            else:
                positive_count += 1

            negate_next = False

        elif word in negative_words:

            if negate_next:
                positive_count += 1
            else:
                negative_count += 1

            negate_next = False

        else:
            negate_next = False

    if positive_count > negative_count:
        return "positive"

    return "negative"


# 5. INTENT DETECTION

def detect_intent(message, tokens):

    scores = {}

    for intent, keywords in topics.items():

        score = 0

        for keyword in keywords:
            if keyword in message:
                score += 2

        for token in tokens:
            if token in keywords:
                score += 1

        scores[intent] = score

    best_intent = max(scores, key=scores.get)

    if scores[best_intent] == 0:
        return None

    return best_intent


# 6. GENERATE RESPONSE

def generate_reply(message):

    tokens = word_tokenize(message.lower())

    sentiment = analyse_sentiment(tokens)

    intent = detect_intent(message.lower(), tokens)

    if intent is None:
        reply = "Sorry, I couldn't understand your message."
    else:
        reply = responses[intent][sentiment]

    return reply, sentiment, intent


# 7. START CHATBOT

print("NLP Chatbot Started!")
print("Ask me about Python, NLP or chatbots.")
print("Type 'bye' to exit.\n")

while True:

    user_input = input("You: ").strip()

    if user_input.lower() in ["bye", "exit", "quit"]:
        print("Bot: Goodbye!")
        break

    if not user_input:
        print("Bot: Please enter a message.")
        continue

    response, sentiment, intent = generate_reply(user_input)

    print("Bot:", response)
    print("Sentiment:", sentiment)
    print()
```
